```python
queue = []

while True:
    print("1. Add a value to the Queue")
    print("2. Remove the value from the Queue")
    print("3. Display the values from Queue")
    print("4. Quit")
    ch = input("Enter your option:")

    if ch=="1":
        val = input("Enter the value to be added: ")
        queue.append(val)
    elif ch=="2":
        if len(queue)<1:
            print("Empty List!")
            continue
        queue.pop(0)
        
    elif ch=="3":
        if len(queue)<1:
            print("Empty List!")
            continue  # continue will take you to the beginning of the loop
        for i in queue:
            print(i, end=" - ")
        print()
    elif ch=="4":
        print("Good Day")
        break
    else:
        print("Invalid option, Try again...")
```

```python
d1 = {}
print(type(d1))   # <class 'dict'>
d1 = {1:"One","Two":2,"Three":"Three","hobbies":["Cricket","Reading","Coding"]}
print(d1)

print(d1['hobbies'])
print(d1.get('hobbies'))

d2 = d1   # Shallow copy - looks like creating a copy but its not creating any copy
d3 = d1.copy()  # Deep copy - creates a new dictionary

d2.clear()
print(d1)
```

```python
d1 = {1:"One","Two":2,"Three":"Three","hobbies":["Cricket","Reading","Coding"]}
print("Dictionary Keys")
print(d1.keys())
for k in d1.keys():
    print(k," - ",d1[k])

print("Dictionary Items (key & value)")  # list of key value tuple [(),()]
print(d1.items())
for pairs in d1.items():
    print(pairs[0]," - ", pairs[1])
print("===")
for k,v in d1.items():
    print(k," - ",v)

print("Dictionary values:")
print(d1.values())
for v in d1.values():
    print(v)
```

```python
all_marks={}
n_st,n_sb = 3,3

for i in range(n_st):
    name = input("Enter your name: ")
    print("Enter the marks for ",name,":")
    temp_list = []
    for j in range(n_sb):
        marks = int(input("Enter the marks in Subject"+str(j+1)+": "))
        temp_list.append(marks)
    temp_list.append(sum(temp_list))
    all_marks.update({name:temp_list})
print("Details:\n",all_marks)
```

```python
all_marks=  {'Sachin': [88, 99, 66, 253], 
             'Rohit': [77, 59, 79, 253], 
             'Virat': [81, 71, 61, 253]}
highest = -1
toppers = []
for k,v in all_marks.items():
    if highest < v[3]:
        highest = v[3]
        toppers.clear()
        toppers.append(k)
    elif highest == v[3]:
        toppers.append(k)
print("Highest marks = ",highest, "scored by",toppers)
```

```python
# Properties of SETS
s1 = {1,3,5,7,9}
s2 = {3,9,5,12,6}

# union, intersection, different, symmetric_difference
print(s1.union(s2))  # update()
print(s1.intersection(s2))  # intersection_update()
print(s1.difference(s2))  # difference_update()
print(s2.difference(s1))
print(s1.symmetric_difference(s2)) #symmetric_difference_update()
print(s1.isdisjoint(s2))
print(s1.issuperset(s1))
print(s1.issubset(s1))

print(s1.issuperset(s2))
print(s1.issubset(s2))
```
