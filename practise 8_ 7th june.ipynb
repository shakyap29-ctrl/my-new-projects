```python
#WAF to check if a number if Prime or not
def my_check_prime(val):
    '''
    Docstring which is used to give documentation to a function. Its a multiline string given as the
    first line in the function.
    This function takes a value and returns True if its prime and False otherwise.
    input argument:
    val: takes input value
    '''
    isPrime = True
    for i in range(2,val//2+1):
        if val % i ==0:
            isPrime = False
            break
    return isPrime
```

```python
val = int(input("Enter a number: "))
isPrime = my_check_prime(val)
if isPrime:
        print(f"{val} is a Prime number")
else:
        print(f"{val} is not a Prime number")
```

```python
# WAP to generate prime numbers between 2 values
print("Prime numbers are:")
for i in range(1000,2000):
    isPrime = my_check_prime(i)
    if isPrime:
        print(i, end=", ")
print()
```

```python
# Special List functions:
# map: list of values where you need to apply same logic over each of the members of the list
# list of USD values and now you want to convert into INR
l1 = [23,45,32,45,35,98]
l2 = []
for i in l1:
    l2.append(i*90.25)

print(l2)
```

```python
l1 = [23,45,32,45,35,98]
l2 = list(map(lambda x:x*90.25, l1))
print(l2)
```

```python
# filter: you want to filter our subset from the main list based on certain criteria
l1 = [23,45,32,45,35,98,100,104,115]
# l2= [45,45,35,100,115]
l2 = list(filter(lambda x: x%5==0, l1))
print(l2)
```

```python
# sorted
names=["Suryavanshi","Rohit","Sachin","Bhuvaneshwar","Pant","Om"]
result = sorted(names, key=lambda x: len(x))
print("result: ",result)
```

```python
# reduce: [1,2,3,4,5], * : logic is applied to all the values and one single answer is calculated
import functools as ft
l1 = [2075.75, 4061.25, 2888.0, 4061.25, 3158.75, 8844.5]
result = ft.reduce(lambda x,y: x*y, l1)
print("Result : ",result)
```

```python
# Single line for loop
l1 = [4,5,12,18,20]
result = [x*x for x in l1]
print("Result = ",result)
```
