```python
#WAP to find average marks obtained by a student in 3 subjects and based on that assign grade,
# rule is: 90-100: A1 80-90: A2, 70-80: B1, 60-70: B2  50-60: C1  40-50: C2  <40: E
m1 = int(input("Enter the marks in subject 1: "))
m2 = int(input("Enter the marks in subject 2: "))
m3 = int(input("Enter the marks in subject 3: "))
avg = (m1 + m2 + m3)/3
print("Average marks obtained is: ",avg)

if avg >=90:
    print("Grade: A1")
elif avg >=80:
    print("Grade: A2")
elif avg >=70:
    print("Grade: B1")
elif avg >=60:
    print("Grade: B2")
elif avg >=50:
    print("Grade: C1")
elif avg>=40:
    print("Grade: C2")
else:
    print("Grade: E")
```

Calculating Electricity Bill Amount:

Unit Consumption - Charges <50 - Free 50-100: Rs 1.50/unit 100-200 : Rs 2.50/unit 200-400: Rs 3.50/unit 400: Rs 5/unit

320: (50)0 + 501.5 + 1002.50 + 1203.50 = 0 + 75 + 250 + 420 = 745

```python
units = int(input("Enter your total units consumed for the month:"))
if units <=50:
    amount = 0
elif units <=100:
    amount = (units-50) * 1.50
elif units <= 200:
    amount = 50*1.5 + (units-100) * 2.50
elif units <= 400:
    amount = 50*1.5 + 100*2.5 + (units-200) * 3.50
else:
    amount = 50*1.5 + 100*2.5 + 200*3.50 + (units-400) * 5.0

print("Total amount payable is: ",amount)
```

```python
#WAP to find average marks obtained by a student in n subjects and based on that assign grade,
# rule is: 90-100: A1 80-90: A2, 70-80: B1, 60-70: B2  50-60: C1  40-50: C2  <40: E
# >95 - merit letter
# repeat it for all the students - when you know in the beginning how many students

n_students = int(input("Enter the number of students:"))
n = int(input("Enter the number of subject: "))

for j in range(n_students):
    print("="*40)
    print("GETTING DATA FOR STUDENT",(j+1),":")
    print("="*40)
    total = 0
    for i in range(n):
        m1 = int(input("Enter the marks in subject "+ str(i+1) +": "))
        total = total + m1
        
    avg = total/n
    print("Average marks obtained is: ",avg)
    
    if avg >=90:
        print("Grade: A1")
        if avg >=95:
            print("You get a Merit Letter")
    elif avg >=80:
        print("Grade: A2")
    elif avg >=70:
        print("Grade: B1")
    elif avg >=60:
        print("Grade: B2")
    elif avg >=50:
        print("Grade: C1")
    elif avg>=40:
        print("Grade: C2")
    else:
        print("Grade: E")
```
