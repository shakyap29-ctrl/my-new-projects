```python
# Arithmetic operators: Mathematical: +  -   *  /    ** {power)   // (integer division)  % Mod (remainder)
var1,var2 = 50,6
print(var1 + var2)
print(var1 - var2)
print(var1 / var2)
print(var1 * var2)
print(var1 ** var2)
print(var1 // var2)
print(99999 // 10000)
print(99999 % 10000)
print(var1 % var2)
```

```python
# relational/camparison (equality):  >   >=  <   <=   == (is it equal to)  != (is not equal to)
# input is values and output is always bool values
var1,var2,var3 = 10,5,10
print(var1 > var2)   #  T
print(var1 > var3)  # F
print(var1 >= var3)  #T

print(var1 < var2)   # F
print(var1 < var3)  # F
print(var1 <= var3)  # T

print(var1 == var2)   # F
print(var1 == var3)  # T

print(var1 != var2)   # T
print(var1 != var3)  # F
```

```python
# logical operator: AND  OR NOT
# input and output are both bool values
#AND: all conditions have to be True for the result to be True

# OR: any one condition needs to be true for the result to be True
print(True and True) # T
print(True and False) #F
print(False and True) #F
print(False and False) #F

print(True or True) # T
print(True or False) # T
print(False or True) # T
print(False or False)  #F

print(not True)
print(not False)
```

```python
# Control flow: we are controlling how program should flow - program will not follow sequence
# conditions and loops
# CONDITIONS
# value1 = 10 is this positive or negative number?

value1 = int(input("Enter a number: "))
# is this +ve or -ve?
# HANDLING JUST ONE CONDITION AT A TIME
if value1 >0:
    print("Positive")
    print("Its a positive number")

# HANDLING TWO CONDITIONS AT A TIME
if value1 >0:
    print("2. Positive")
else:
    print("Not Positive")

# HANDLING MULTIPLE CONDITIONS (> 2) AT A TIME:  if  elif...  else
if value1 > 0:
    print("3. Positive")
elif value1 <0:
    print("3. Negative")
else:
    print("3. Zero is neither positive nor negative")
    
print("Thank You")
```
