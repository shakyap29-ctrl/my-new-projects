```python
#WAP to calculate profit given the cost price and the sales price
# take the input from the user: input() - it can accept maximum of 1 parameter (str)
# by default input() reads value as string.
# int()  float() complex() bool() str()

cost_price = input("Enter the Cost Price: ")
print(type(cost_price))
cost_price = float(cost_price)
sales_price = float(input("Enter the Sales Price: "))
profit = sales_price-cost_price
print("The profit made after buying a product for Rs",cost_price,"and selling it at Rs", sales_price,"is Rs",profit)
# f-string (format string)
print(f"The profit made after buying a product for Rs {cost_price} and selling it at Rs {sales_price} is Rs {profit:.5f}")
```
