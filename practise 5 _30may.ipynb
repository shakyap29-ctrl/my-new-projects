```python
# Develop guess the number game
import random
count = 0
max_life = 2
curr_life = 0
num = random.randint(1,100)  # generates a random integer number between given 2 number both inclusive
while True:
    guess = int(input("Guess the number(1-100): "))
    if guess >100 or guess <0:
        if curr_life == max_life - 1:
            print("You have lost all your lives!!!!")
            break
        else:
            curr_life+=1
            print("Invalid guess. You lose a life. Total Life Remaining =",max_life -curr_life)
            continue  # will take you to the beginning of the loop
            
    count+=1  # count = count + 1 - > a += b
    if guess==num:
        print(f"Congratulations! You win with {count} attempts :)")
        break
    else:
        if guess>num:
            print("Incorrect! Try Again with a lower number...")
        else:
            print("Incorrect! Try Again with a higher number...")
```

```python
'''
In python counting (indexing) happens both ways - left to right and right to left

0  1   2   3   4    L to R
H  E   L   L   O
-5 -4  -3  -2  -1   R to L

'''
str1 = "HELLO"
print("1st character: ",str1[0], str1[-5], str1[len(str1)-5])
print("last character: ",str1[len(str1)-1], str1[-1], str1[4])

# Strings are immutable: cant be edited
#str1[0] = "K"  TypeError: 'str' object does not support item assignment - cant assign a value
str1 = "K" + str1[1:5]   #just like range here too we need to give +1 for ending
print(str1)
print("KELL using -ve indexing: ",str1[-5:-1])
print("LLO using -ve indexing: ",str1[-3:])
```
