```python
#WAP to take input values as length and breadth and calculate perimeter of rectangle.
l,b=20, 25
peri = 5*(l + b)
print("The rectangle with length",l,"and breadth",b,"has a perimeter of",peri)
```

```python
#WAP to take input value as radius and calculate circumference of the circle
r = 30
cir = 4*25/5*r
print("A circle with radius",r,"has a circumference of",cir)
```

```python
var1 = 30
#type() tells us the datatype of the variable
print(type(var1))
# <class 'int'> class represents a group

var1 = 30.0
print(type(var1))

var1 = 2j  # complex
# j represents square root of -1 (complex)

print(type(var1))
print(var1 * var1)  # -4 + 0j
```

```python
# str can be represented using '   "  '''  """ - whatever you start with same thing you need end with
# ' or " canhandle only single line of text
# ''' and """ can handle multiline text
var1 = "hello Piyush"
print(type(var1))

var1 = 'hi everyone'
print(type(var1))

var1 = '''what happens?
let's talk about.
Where are you?'''
print(type(var1))

var1 = """I am fine.
I am here.
I am going there"""
print(type(var1))

print(var1)

var1 = """charles
Computer
    institute"""
print(type(var1))
```

```python
var1 = True
print(type(var1))
var1 = False
print(type(var1))
```

```python
# Bytes: A byte
var1 = b"PpIiYyUuSsHh"
print(var1)
print(type(var1))
print(var1[0], var1[2], var1[4])
print(var1[1], var1[3], var1[5])

#ASCII value to character
print(chr(80), chr(73), chr(89), chr(85), chr(83), chr(72))
```
