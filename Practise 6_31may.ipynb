```python
str1 = "I am FINE How FIne are YOU Hope Fine is finE for finery all"
# count() find() index()
substr = "fine"
if substr.upper() in str1.upper():
    print("Yes we do have")
else:
    print("No not in the string")

cnt = str1.lower().count(substr)
print(f"There are {cnt} occurrences of the substring")

# position (index) of the first occurrence
# but incase substr is not there then - find will return -1 but index will give error
print(str1.lower().find("fined"))
print(str1.lower().index("fine"))

print(str1.lower().find("fine",10,20))

#Program to find position of all the occurrences of the substr
# 5, (pos+1) 14, (15..) 29, 
st=0
print("The substr is in the main string at the following positions: ",end=" ")
for i in range(str1.lower().count(substr)):
    pos = str1.lower().find(substr,st)
    print(pos,end=", ")
    st = pos+1
```

```python
# strip() "    this is        fine   "  rstrip()  lstrip() beginningwith() endingwith()
# split()  join()
str1 = "I am FINE How FIne are YOU Hope Fine is finE for finery all"
result = str1.split()
print("1 Split = ",result)
str2 = " - ".join(result)
print("STR 2: ",str2)

result = str1.lower().split("fine")
print("1 Split = ",result)
str2 = "fine".join(result)
print("STR 2: ",str2)
```

```python
l2 = [55,"Hello",3j, True, 5.6, [1,2,3]]
# 3 ways to add: insert(pos,value)  append(value) extend(list)
print("0. List 2 = ",l2)
l2.insert(2,56)
l2.insert(2,"Piyush")
print("1. List 2 = ",l2)
l2.append(21)
print("2. List 2 = ",l2)

l2.extend([41,51,61])  # l1 = l1 + l2
print("3. List 2 = ",l2)

print(l2.count(5.6))
# delete the values from the list:  remove(value) pop(pos)
l2.remove(5.6)  # first check with count if value in the list
print("4. List 2 = ",l2)
l2.pop(-2) #check the index value with len()
print("5. List 2 = ",l2)
# I want to remove the position where value 21 is there
# index that gives position
l2.pop(l2.index(21))
print("6. List 2 = ",l2)
```

```python
#creating list from a list
# = they are creating different names for the same list - list is only one
# .copy() - created a new list (xerox)
l3 = l2
l4 = l2
l5 = l2.copy()
print("7. List 2 = ",l2)
print("7. List 3 = ",l3)
print("7. List 4 = ",l4)
print("7. List 5 = ",l5)
l2.append(77)
l3.append(88)
l4.append(99)
l5.append(111)
print("8. List 2 = ",l2)
print("8. List 3 = ",l3)
print("8. List 4 = ",l4)
print("8. List 5 = ",l5)
```

```python
l2 =  [55, 56, 41, 61, 77, 88, 99, 77, 88, 99]
l2.reverse()
print("1. L2 = ",l2)
l2.sort()  # ASC
print("2. L2 = ",l2)
l2.sort(reverse=True)  # DESC
print("3. L2 = ",l2)
l2.clear()
print("4. L2 = ",l2)
```

```python
# read marks of 3 students in 3 subjects and then check who has scored highest in each subject
# read the marks
n_st = 3
n_sb = 3
all_marks = []
for i in range(n_st):
    print(f"Enter the marks for student {i+1} : ")
    t_list = []
    for j in range(n_sb):
        marks = int(input("Enter the marks in subject "+str(j+1)+" :"))
        t_list.append(marks)
    all_marks.append(t_list)

print("All marks are: ",all_marks)
```

```python
#all_marks
student_idx = []

for marks in tuple(all_marks):
    high_marks = -1
    for i in range(n_st):
        #print(marks[i])
        if high_marks < marks[i]:
            high_marks = marks[i]
            idx = i
    student_idx.append(idx)

print(student_idx)

for i in range(len(student_idx)):
    print(f"Highest marks in subject {i+1} is obtained by Student {student_idx[i]+1}")
```
